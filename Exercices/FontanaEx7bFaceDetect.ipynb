# Fontana Emanuele, Haar Face Detector #

```python
import cv2
import matplotlib.pyplot as plt
```

```python
img=cv2.cvtColor(cv2.imread('Faces.png'),cv2.COLOR_BGR2RGB)
plt.imshow(img,vmin=0,vmax=255)
plt.axis('off')
plt.title('Original Image')
plt.show()

grayImg=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
plt.imshow(grayImg,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Grayscale Image')
plt.show()
```

## Now let's apply di Face Detector ##

```python
faces_cascade=cv2.CascadeClassifier(cv2.data.haarcascades+'haarcascade_frontalface_default.xml')
faces=faces_cascade.detectMultiScale(grayImg,scaleFactor=1.1,minNeighbors=1)
copy=grayImg.copy()
for (x,y,w,h) in faces:
    cv2.rectangle(copy,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copy,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces')
plt.show()
```

## Let's apply some transformation to our photo ##

```python
rotated90_image=cv2.rotate(grayImg,cv2.ROTATE_90_CLOCKWISE)
rotated180_image=cv2.rotate(grayImg,cv2.ROTATE_180)
equalized_image=cv2.equalizeHist(grayImg)
blurred_image=cv2.GaussianBlur(grayImg,(5,5),0)
double_size_image=cv2.resize(grayImg,(0,0),fx=2,fy=2)
half_size_image=cv2.resize(grayImg,(0,0),fx=0.5,fy=0.5)

plt.figure(figsize=(10, 8))
plt.subplot(2, 2, 1)
plt.imshow(rotated90_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Rotated 90 degrees')

plt.subplot(2, 2, 2)
plt.imshow(rotated180_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Rotated 180 degrees')

plt.subplot(2, 2, 3)
plt.imshow(equalized_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Equalized Image')

plt.subplot(2, 2, 4)
plt.imshow(blurred_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Blurred Image')
plt.tight_layout()
plt.show()


plt.figure(figsize=(10, 5))
plt.imshow(double_size_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Double Size Image')
plt.show()

plt.imshow(half_size_image, cmap='gray', vmin=0, vmax=255)
plt.axis('off')
plt.title('Half Size Image')
plt.tight_layout()
plt.show()
```

## Let's use the face detector again  (use the same arguments values of the original one) ## 

```python
faces90=faces_cascade.detectMultiScale(rotated90_image,scaleFactor=1.1,minNeighbors=1)
copy90=rotated90_image.copy()
for (x,y,w,h) in faces90:
    cv2.rectangle(copy90,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copy90,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Rotated 90 degrees Image')
plt.show()

faces180=faces_cascade.detectMultiScale(rotated180_image,scaleFactor=1.1,minNeighbors=1)
copy180=rotated180_image.copy()
for (x,y,w,h) in faces180:
    cv2.rectangle(copy180,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copy180,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Rotated 180 degrees Image')
plt.show()

facesEqualized=faces_cascade.detectMultiScale(equalized_image,scaleFactor=1.1,minNeighbors=1)
copyEqualized=equalized_image.copy()
for (x,y,w,h) in facesEqualized:
    cv2.rectangle(copyEqualized,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copyEqualized,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Equalized Image')
plt.show()

facesBlurred=faces_cascade.detectMultiScale(blurred_image,scaleFactor=1.1,minNeighbors=1)
copyBlurred=blurred_image.copy()
for (x,y,w,h) in facesBlurred:
    cv2.rectangle(copyBlurred,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copyBlurred,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Blurred Image')
plt.show()

facesDouble=faces_cascade.detectMultiScale(double_size_image,scaleFactor=1.1,minNeighbors=1)
copyDouble=double_size_image.copy()
for (x,y,w,h) in facesDouble:
    cv2.rectangle(copyDouble,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copyDouble,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Double Size Image')
plt.show()

facesHalf=faces_cascade.detectMultiScale(half_size_image,scaleFactor=1.1,minNeighbors=1)
copyHalf=half_size_image.copy()
for (x,y,w,h) in facesHalf:
    cv2.rectangle(copyHalf,(x,y),(x+w,y+h),(255,255,255),2)
plt.imshow(copyHalf,cmap='gray',vmin=0,vmax=255)
plt.axis('off')
plt.title('Detected Faces in Half Size Image')
plt.show()
```

## Comment on the results ##

1. If we rotate our photo by 90° then we are not able to detect face anymore because we are using the same filters
2. By turning upside down the photo, instead, we are detecting almost the same regions
3. Equalizing our photo to improve the contrast, we are able to detect a new face ---> now filters are able to match that portion of the photo as a face because of better intensity level
4. Of course, by blooring our image we are losing details and now the number of detected faces has decreased
5. By doubling the size now we are able to detect a new faces, but we also increase the number of false positive
6. By halfing the size now we are losing some faces but we also reduce the number of false positive

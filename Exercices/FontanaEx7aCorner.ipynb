# Fontana Emanuele, 7a Corner Detection #

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img=cv2.cvtColor(cv2.imread('corner.png'), cv2.COLOR_BGR2GRAY)

plt.imshow(img, cmap='gray',vmin=0, vmax=255)
plt.title('Original Image'), plt.xticks([]), plt.yticks([])
plt.show()

# Harris Corner Detection
Harris = cv2.cornerHarris(img, blockSize=20, ksize=15, k=0.06)
plt.imshow(Harris,cmap='jet')
plt.axis('off')
plt.title('Harris matrix')
plt.show()


# Shi-Tomasi Corner Detection
ShiTomasi = cv2.goodFeaturesToTrack(img, maxCorners=100, qualityLevel=0.01, minDistance=10)
""""
maxCorners: Maximum number of corners to find
qualityLevel: used to filter out corners
minDistance: Minimum distance between detected corners
"""
for i in ShiTomasi:
    x, y = i.ravel()
    x, y = int(x), int(y)  
    cv2.circle(img, (x, y), 3, (0, 255, 0), -1)

plt.imshow(img)
plt.axis('off')
plt.title('Shi-Tomasi corners')
plt.show()
```

We can say that with Shi-Tomasi we are also able to detect "round corners" and not only "sharp corners". Look at the last image for example

# First Photo in grayscale #

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img= cv2.imread('Q2_1_1.tiff')
imgGray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
histGray = cv2.calcHist([imgGray], [0], None, [256], [0, 256])
cdF = histGray.cumsum()
cdF_normalized = cdF * histGray.max() / cdF.max() 

plt.plot(histGray, color='blue', label='Histogram')
plt.plot(cdF_normalized, color='red', label='Normalized CDF')
plt.title('Histogram and CDF of Gray Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.xlim([0, 256])
plt.legend()
plt.show()

equalizedGray = cv2.equalizeHist(imgGray)
fig, ax = plt.subplots(1,2, figsize=(10,5))
ax[0].imshow(imgGray,'gray', vmin=0, vmax=255)
ax[0].set_title('Original', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(equalizedGray,'gray', vmin=0, vmax=255)
ax[1].set_title('Equalized', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])
plt.tight_layout()
plt.show()


equalizedGrayHist = cv2.calcHist([equalizedGray], [0], None, [256], [0, 256])
cdF = equalizedGrayHist.cumsum()
cdF_normalized = cdF * equalizedGrayHist.max() / cdF.max()
plt.plot(equalizedGrayHist, color='blue', label='Histogram')
plt.plot(cdF_normalized, color='red', label='Normalized CDF')
plt.title('Histogram and CDF of Equalized Gray Image')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')
plt.xlim([0, 256])
plt.legend()
plt.show()
```

# First photo with all colors #

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Split the image into the three color channels
b, g, r = cv2.split(img_rgb)

# Apply histogram equalization to each channel
equalized_b = cv2.equalizeHist(b)
equalized_g = cv2.equalizeHist(g)
equalized_r = cv2.equalizeHist(r)

# Merge the equalized channels back together
equalized_rgb = cv2.merge([equalized_r, equalized_g, equalized_b])

# Plot the original and equalized images
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img_rgb,vmin=0, vmax=255)
ax[0].set_title('Original RGB', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])

ax[1].imshow(equalized_rgb,vmin=0, vmax=255)
ax[1].set_title('Equalized RGB', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])

plt.tight_layout()
plt.show()

# Plot histograms and CDFs for each channel
fig, axes = plt.subplots(3, 2, figsize=(12, 10))


hist_r = cv2.calcHist([r], [0], None, [256], [0, 256])
cdf_r = hist_r.cumsum()
cdf_r_normalized = cdf_r * hist_r.max() / cdf_r.max()
axes[0, 0].plot(hist_r, color='red', label='Histogram')
axes[0, 0].plot(cdf_r_normalized, color='black', label='Normalized CDF')
axes[0, 0].set_title('Original Red Channel')
axes[0, 0].set_xlim([0, 256])
axes[0, 0].legend()

hist_r_eq = cv2.calcHist([equalized_r], [0], None, [256], [0, 256])
cdf_r_eq = hist_r_eq.cumsum()
cdf_r_eq_normalized = cdf_r_eq * hist_r_eq.max() / cdf_r_eq.max()
axes[0, 1].plot(hist_r_eq, color='red', label='Histogram')
axes[0, 1].plot(cdf_r_eq_normalized, color='black', label='Normalized CDF')
axes[0, 1].set_title('Equalized Red Channel')
axes[0, 1].set_xlim([0, 256])
axes[0, 1].legend()


hist_g = cv2.calcHist([g], [0], None, [256], [0, 256])
cdf_g = hist_g.cumsum()
cdf_g_normalized = cdf_g * hist_g.max() / cdf_g.max()
axes[1, 0].plot(hist_g, color='green', label='Histogram')
axes[1, 0].plot(cdf_g_normalized, color='black', label='Normalized CDF')
axes[1, 0].set_title('Original Green Channel')
axes[1, 0].set_xlim([0, 256])
axes[1, 0].legend()

hist_g_eq = cv2.calcHist([equalized_g], [0], None, [256], [0, 256])
cdf_g_eq = hist_g_eq.cumsum()
cdf_g_eq_normalized = cdf_g_eq * hist_g_eq.max() / cdf_g_eq.max()
axes[1, 1].plot(hist_g_eq, color='green', label='Histogram')
axes[1, 1].plot(cdf_g_eq_normalized, color='black', label='Normalized CDF')
axes[1, 1].set_title('Equalized Green Channel')
axes[1, 1].set_xlim([0, 256])
axes[1, 1].legend()


hist_b = cv2.calcHist([b], [0], None, [256], [0, 256])
cdf_b = hist_b.cumsum()
cdf_b_normalized = cdf_b * hist_b.max() / cdf_b.max()
axes[2, 0].plot(hist_b, color='blue', label='Histogram')
axes[2, 0].plot(cdf_b_normalized, color='black', label='Normalized CDF')
axes[2, 0].set_title('Original Blue Channel')
axes[2, 0].set_xlim([0, 256])
axes[2, 0].legend()

hist_b_eq = cv2.calcHist([equalized_b], [0], None, [256], [0, 256])
cdf_b_eq = hist_b_eq.cumsum()
cdf_b_eq_normalized = cdf_b_eq * hist_b_eq.max() / cdf_b_eq.max()
axes[2, 1].plot(hist_b_eq, color='blue', label='Histogram') 
axes[2, 1].plot(cdf_b_eq_normalized, color='black', label='Normalized CDF')
axes[2, 1].set_title('Equalized Blue Channel')
axes[2, 1].set_xlim([0, 256])
axes[2, 1].legend()

plt.tight_layout()
plt.show()
```

## Adaptive histogram equalization ##

```python
imgGray = cv2.cvtColor(cv2.imread('Q2_1_1.tiff'), cv2.COLOR_BGR2GRAY)
# Calculate histogram and CDF
histGray = cv2.calcHist([imgGray], [0], None, [256], [0, 256])
cdfGray = histGray.cumsum()
cdfGray_normalized = cdfGray * histGray.max() / cdfGray.max()
equalizer=cv2.createCLAHE(tileGridSize=(8,8))
equalized=equalizer.apply(imgGray)
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(imgGray, 'gray', vmin=0, vmax=255)
ax[0].set_title('Original', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(equalized, 'gray', vmin=0, vmax=255)
ax[1].set_title('Equalized', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])
plt.tight_layout()
plt.show()
equalizedHist = cv2.calcHist([equalized], [0], None, [256], [0, 256])
cdf = equalizedHist.cumsum()
cdf_normalized = cdf * equalizedHist.max() / cdf.max()
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# Plot histogram and CDF for original image
axes[0].plot(histGray, color='blue', label='Histogram')
axes[0].plot(cdfGray_normalized, color='red', label='Normalized CDF')
axes[0].set_title('Histogram and CDF of Original Image')
axes[0].set_xlabel('Pixel Value')
axes[0].set_ylabel('Frequency')
axes[0].set_xlim([0, 256])
axes[0].legend()
# Plot histogram and CDF for equalized image
axes[1].plot(equalizedHist, color='blue', label='Histogram')
axes[1].plot(cdf_normalized, color='red', label='Normalized CDF')
axes[1].set_title('Histogram and CDF of Equalized Image')
axes[1].set_xlabel('Pixel Value')
axes[1].set_ylabel('Frequency')
axes[1].set_xlim([0, 256])
axes[1].legend()
plt.tight_layout()
plt.show()
```

# Linear stretching of the first photo #

```python
img=cv2.cvtColor(cv2.imread('Q2_1_1.tiff'), cv2.COLOR_BGR2GRAY)
a=img.min()
b=img.max()
lookUpTable = np.empty((1, 256), dtype='uint8')
for i in range(256):
    lookUpTable[0,i] = np.clip(255*((i-a)/(b-a)), 0, 255)
processedImage = cv2.LUT(img, lookUpTable)
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(img, 'gray', vmin=0, vmax=255)
ax[0].set_title('Original', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(processedImage, 'gray', vmin=0, vmax=255)
ax[1].set_title('Processed', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])
plt.tight_layout()
plt.show()
# Calculate histogram and CDF for processed image
processedHist = cv2.calcHist([processedImage], [0], None, [256], [0, 256])
cdf = processedHist.cumsum()
cdf_normalized = cdf * processedHist.max() / cdf.max()
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# Plot histogram and CDF for original image
axes[0].plot(histGray, color='blue', label='Histogram')
axes[0].plot(cdfGray_normalized, color='red', label='Normalized CDF')
axes[0].set_title('Histogram and CDF of Original Image')
axes[0].set_xlabel('Pixel Value')
axes[0].set_ylabel('Frequency')
axes[0].set_xlim([0, 256])
axes[0].legend()
# Plot histogram and CDF for processed image
axes[1].plot(processedHist, color='blue', label='Histogram')
axes[1].plot(cdf_normalized, color='red', label='Normalized CDF')
axes[1].set_title('Histogram and CDF of Processed Image')
axes[1].set_xlabel('Pixel Value')
axes[1].set_ylabel('Frequency')
axes[1].set_xlim([0, 256])
axes[1].legend()
plt.tight_layout()
plt.show()
```

## Part 2: Contrast enhancement of color image ##

```python
img= cv2.imread('Q2_1_1.tiff')
HSVimg = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
H, S, V = cv2.split(HSVimg)
a=V.min()
b=V.max()
lookUpTable = np.empty((1, 256), dtype='uint8')
for i in range(256):
    lookUpTable[0,i] = np.clip(255*((i-a)/(b-a)), 0, 255)
processedV = cv2.LUT(V, lookUpTable)
processedHSV = cv2.merge([H, S, processedV])
#plot original and processed image
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB), vmin=0, vmax=255)
ax[0].set_title('Original', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(cv2.cvtColor(cv2.merge([H, S, processedV]), cv2.COLOR_HSV2RGB), vmin=0, vmax=255)
ax[1].set_title('Processed', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])
plt.tight_layout()
plt.show()


img2= cv2.imread('Q2_1_2.tiff')
HSVimg2 = cv2.cvtColor(img2, cv2.COLOR_BGR2HSV)
H2, S2, V2 = cv2.split(HSVimg2)
a=V2.min()
b=V2.max()
lookUpTable = np.empty((1, 256), dtype='uint8')
for i in range(256):
    lookUpTable[0,i] = np.clip(255*((i-a)/(b-a)), 0, 255)
processedV2 = cv2.LUT(V2, lookUpTable)
processedHSV2 = cv2.merge([H2, S2, processedV2])
#plot original and processed image
fig, ax = plt.subplots(1, 2, figsize=(10, 5))
ax[0].imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB), vmin=0, vmax=255)
ax[0].set_title('Original', fontsize=18, color='gray')
ax[0].set_xticks([])
ax[0].set_yticks([])
ax[1].imshow(cv2.cvtColor(cv2.merge([H2, S2, processedV2]), cv2.COLOR_HSV2RGB), vmin=0, vmax=255)
ax[1].set_title('Processed', fontsize=18)
ax[1].set_xticks([])
ax[1].set_yticks([])
plt.tight_layout()
plt.show()
```

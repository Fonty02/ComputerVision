# Fontana Emanuele 7a, HOG Descriptor #

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img1=cv2.cvtColor(cv2.imread('man-face.jpeg'), cv2.COLOR_BGR2GRAY)
img2=cv2.cvtColor(cv2.imread('man-face2.jpeg'), cv2.COLOR_BGR2GRAY)

plt.subplot(121), plt.imshow(img1, cmap='gray',vmin=0, vmax=255)
plt.title('Image 1'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(img2, cmap='gray',vmin=0, vmax=255)
plt.title('Image 2'), plt.xticks([]), plt.yticks([])
plt.show()
```

Resize of the images

```python
img1=cv2.resize(img1, (512, 512))
img2=cv2.resize(img2, (512, 512))

plt.subplot(121), plt.imshow(img1, cmap='gray',vmin=0, vmax=255)
plt.title('Image 1'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(img2, cmap='gray',vmin=0, vmax=255)
plt.title('Image 2'), plt.xticks([]), plt.yticks([])
plt.show()
```

Compute the HOG

```python
hog1 = cv2.HOGDescriptor()
hog2 = cv2.HOGDescriptor()
hog_features1 = hog1.compute(img1)
hog_features2 = hog2.compute(img2)
plt.figure(figsize=(12, 6))


plt.subplot(1, 2, 1)
plt.imshow(img1, cmap='gray')
plt.title("Immagine 1 - HOG")
plt.axis('off')


plt.subplot(1, 2, 2)
plt.imshow(img2, cmap='gray')
plt.title("Immagine 2 - HOG")
plt.axis('off')

plt.show()
```

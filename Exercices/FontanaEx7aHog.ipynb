# Fontana Emanuele 7a, HOG Descriptor #

```python
import cv2
from skimage.feature import hog
import matplotlib.pyplot as plt
import numpy as np
```

```python
img1=cv2.cvtColor(cv2.imread('Hank.png'), cv2.COLOR_BGR2GRAY)
img2=cv2.cvtColor(cv2.imread('Jesse.png'), cv2.COLOR_BGR2GRAY)

plt.subplot(121), plt.imshow(img1, cmap='gray',vmin=0, vmax=255)
plt.title('Image 1'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(img2, cmap='gray',vmin=0, vmax=255)
plt.title('Image 2'), plt.xticks([]), plt.yticks([])
plt.show()
```

Resize of the images

```python
img1=cv2.resize(img1, (512, 512))
img2=cv2.resize(img2, (512, 512))

plt.subplot(121), plt.imshow(img1, cmap='gray',vmin=0, vmax=255)
plt.title('Image 1'), plt.xticks([]), plt.yticks([])
plt.subplot(122), plt.imshow(img2, cmap='gray',vmin=0, vmax=255)
plt.title('Image 2'), plt.xticks([]), plt.yticks([])
plt.show()
```

Compute the HOG

```python
fd_l1, hog1 = hog(img1, orientations=10, pixels_per_cell=(8, 8), cells_per_block=(4, 4), visualize=True)
fd_l2, hog2 = hog(img2, orientations=10, pixels_per_cell=(8, 8), cells_per_block=(4, 4), visualize=True)
cosine_similarity = fd_l1 @ fd_l2 / (np.linalg.norm(fd_l1) * np.linalg.norm(fd_l2))


fig, axs = plt.subplots(nrows = 1, ncols = 2, figsize = (30, 30))

axs[0].imshow(hog1, 'gray')
axs[0].set_title("Hank HOG", fontsize=20)
axs[0].set_xticks([])
axs[0].set_yticks([])

axs[1].imshow(hog2, 'gray')
axs[1].set_title("Jesse HOG", fontsize=20)
axs[1].set_xticks([])
axs[1].set_yticks([])

fig.suptitle(f"Cosine Similarity: {cosine_similarity:.2f}", fontsize=30)
plt.subplots_adjust(wspace=0.1, hspace=0.0000000000000001)
plt.show()
```

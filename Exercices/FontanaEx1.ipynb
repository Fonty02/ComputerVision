## Import of the required libraries ##

## Reading and visualizing the image ##

```python
img=cv2.imread('Dog.png')
# Convert the image from BGR to RGB, as OpenCV loads images in BGR format
img_rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
width, height = img.shape[1], img.shape[0]
plt.imshow(img_rgb,vmin=0, vmax=255)
plt.title('RGB Dog')
plt.xlabel('Width: ' + str(width) + ' pixels')
plt.ylabel('Height: ' + str(height) + ' pixels')
plt.show()


# Actually, OpenCV has a built-in function to show images, but it cannot be used in Jupyter Notebook.
#cv2.imshow('Dog', img)
```

## Let's scale the image ##

```python
scale_percent = 25
width = int(img.shape[1] * scale_percent / 100) #columns
height = int(img.shape[0] * scale_percent / 100) #rows
dim = (width, height)
resized = cv2.resize(img, dim, interpolation = cv2.INTER_AREA)
# cv2.INTER_AREA is used for shrinking an image, which means it uses pixel area relation. It resamples the image by averaging pixel values in the area of the pixel being resized
resized_rgb=cv2.cvtColor(resized,cv2.COLOR_BGR2RGB)
plt.imshow(resized_rgb,vmin=0, vmax=255)
plt.title('Resized 25% Dog')
plt.xlabel(f'Width: {resized.shape[1]} pixels')
plt.ylabel(f'Height: {resized.shape[0]} pixels')
plt.show()

scale_percent = 50
width = int(img.shape[1] * scale_percent / 100) #columns
height = int(img.shape[0] * scale_percent / 100) #rows
dim = (width, height)
resized = cv2.resize(img, dim, interpolation = cv2.INTER_AREA)
# cv2.INTER_AREA is used for shrinking an image, which means it uses pixel area relation. It resamples the image by averaging pixel values in the area of the pixel being resized
resized_rgb=cv2.cvtColor(resized,cv2.COLOR_BGR2RGB)
plt.imshow(resized_rgb,vmin=0, vmax=255)
plt.title('Resized 50% Dog')
plt.xlabel(f'Width: {resized.shape[1]} pixels')
plt.ylabel(f'Height: {resized.shape[0]} pixels')
plt.show()


scale_percent = 75
width = int(img.shape[1] * scale_percent / 100) #columns
height = int(img.shape[0] * scale_percent / 100) #rows
dim = (width, height)
resized = cv2.resize(img, dim, interpolation = cv2.INTER_AREA)
# cv2.INTER_AREA is used for shrinking an image, which means it uses pixel area relation. It resamples the image by averaging pixel values in the area of the pixel being resized
resized_rgb=cv2.cvtColor(resized,cv2.COLOR_BGR2RGB)
plt.imshow(resized_rgb,vmin=0, vmax=255)
plt.title('Resized 75% Dog')
plt.xlabel(f'Width: {resized.shape[1]} pixels')
plt.ylabel(f'Height: {resized.shape[0]} pixels')
plt.show()
```

### As we can see, the less the value of the scaling factor, the more details we are losing. ###

## Let's visualize the image in HSV and Lab color spaces ##

```python
img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
img_lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
fig,(ax1,ax2)=plt.subplots(nrows=1,ncols=2,figsize=(30,30))
ax1.imshow(img_hsv,vmin=0, vmax=255)
ax1.set_title('HSV Dog')
ax2.imshow(img_lab,vmin=0, vmax=255)
ax2.set_title('LAB Dog')
plt.show()
```

### The HSV separates the image into three components: Hue, Saturation, and Value (Brightness). The Hue component represents the color type, the Saturation component represents the intensity of the color, and the Value component represents the brightness of the color. This separation allows for more intuitive manipulation of colors in images. ###

### The LAB is based on the theory of opponent colors. It has 3 channels: L (lightness), A (green-red), and B (blue-yellow). The L channel represents the lightness of the color, while the A and B channels represent the color information. ###

 ### So of course, when we use matplotlib to visualize the image, we are seeing strange colors because we are not using the corrrct color space (which is RGB). The same applies to the Lab color space. ###

## Let's now visualize Green, Red and blue channels separately ##

```python
(B,G,R)=cv2.split(img) #function to split the image into its three channels
fig, axes = plt.subplots(nrows=2, ncols=3, figsize=(50, 50))

# Flatten the 2D array of axes for easier indexing
ax1, ax2, ax3, ax4, ax5, ax6 = axes.flatten()

# Display individual RGB channels
ax1.imshow(R, cmap='Reds', vmin=0, vmax=255)
ax1.set_title('Red Channel', fontsize=100)

ax2.imshow(G, cmap='Greens', vmin=0, vmax=255)
ax2.set_title('Green Channel', fontsize=100)

ax3.imshow(B, cmap='Blues', vmin=0, vmax=255)
ax3.set_title('Blue Channel', fontsize=100)

# Convert each channel to grayscale and display
ax4.imshow(R, cmap='gray', vmin=0, vmax=255)
ax4.set_title('Red Channel in Grayscale', fontsize=50)

ax5.imshow(G, cmap='gray', vmin=0, vmax=255)
ax5.set_title('Green Channel in Grayscale', fontsize=50)

ax6.imshow(B, cmap='gray', vmin=0, vmax=255)
ax6.set_title('Blue Channel in Grayscale', fontsize=50)

plt.show()
```

### The background is more influenced by red and green. The dog's fur has a strong presence of blue and green. ###
